# Basic

```python
povrsini = [54.5, 68, 72.3, 85, 61.2]
print(povrsini)
```

```python
vkupno = 0
for p in povrsini:
    vkupno += p
print("Вкупна површина:", vkupno, "m²")
```

```python
def prosek(lista):
    return sum(lista) / len(lista)

avg = prosek(povrsini)
print("Просек:", avg)
```

```python
if avg > 70:
    print(f"Просекот ({avg:.2f} m²) е над 70 m²")
else:
    print(f"Просекот ({avg:.2f} m²) не е над 70 m²")
```

#Advanced

```python
def statistika(lista):
    return min(lista), max(lista), sum(lista) / len(lista)

mn, mx, avg = statistika(povrsini)
print(f"Мин: {mn}, Макс: {mx}, Просек: {avg:.2f}")
```

```python
def normaliziraj(lista):
    mn, mx = min(lista), max(lista)
    return [(x - mn) / (mx - mn) for x in lista]

print(normaliziraj(povrsini))
```

```python
for i, v in enumerate(povrsini):
    print(f"Индекс {i}: {v} m²")
```

```python
klasi = {
    1: "мал стан",
    2: "среден стан",
    3: "голем стан",
    4: "пентхаус"
}

def klasa(broj):
    return klasi.get(broj, "непозната")

print(klasa(2))   # среден стан
print(klasa(9))   # непозната
```
